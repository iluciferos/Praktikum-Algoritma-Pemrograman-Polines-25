## 📊 Praktikum 02 — Membaca Data CSV dengan Pandas
**Tujuan:** Membaca file CSV menggunakan `pd.read_csv()` dan inspeksi awal DataFrame.

Fungsi yang dipakai:

- `.shape` → jumlah baris & kolom

````python
# ============================================================
# PRAKTIKUM 02: Membaca Data CSV dengan Pandas
# ============================================================

import pandas as pd

NAMA_FILE_CSV = 'lokasi_semarang.csv'


def baca_data_lokasi(nama_file: str):
    """
    Membaca data lokasi dari file CSV menggunakan Pandas.
    Mengembalikan DataFrame jika berhasil, None jika gagal.
    """
    print(f'Mencoba membaca file CSV: {nama_file}')
    try:
        dataframe = pd.read_csv(nama_file)
        print(' -> File CSV berhasil dibaca.')
        return dataframe
    except FileNotFoundError:
        print(f' -> ERROR: File "{nama_file}" tidak ditemukan!')
        return None
    except pd.errors.EmptyDataError:
        print(f' -> ERROR: File "{nama_file}" kosong.')
        return None
    except Exception as e:
        print(f' -> ERROR saat membaca file CSV: {type(e).__name__} - {e}')
        return None


# --- Kode Utama ---
print('--- Praktikum 2: Membaca CSV ---')
df_lokasi = baca_data_lokasi(NAMA_FILE_CSV)

if df_lokasi is not None:
    print('\n--- Inspeksi Awal DataFrame ---')

    print('\n1. Lima Baris Pertama (head()):')
    print(df_lokasi.head())

    print('\n2. Informasi DataFrame (info()):')
    df_lokasi.info()

    jumlah_baris, jumlah_kolom = df_lokasi.shape
    print(f'\n3. Dimensi Data:')
    print(f'   Jumlah Lokasi (Baris) : {jumlah_baris}')
    print(f'   Jumlah Atribut (Kolom): {jumlah_kolom}')

    print(f'\n4. Nama Kolom:')
    print(list(df_lokasi.columns))
else:
    print('\nTidak dapat melanjutkan karena gagal membaca CSV.')

print('\n--- Praktikum 2 Selesai ---')
````

## 🏗️ Praktikum 03 — Mendesain dan Membuat Kelas OOP
**Tujuan:** Membuat hierarki kelas untuk data geografis.

Struktur kelas:
```
Lokasi (Abstract)
├── TempatWisata
├── Kuliner
└── TempatIbadah
```
- `Lokasi` → kelas abstrak (ABC), punya metode abstrak `get_info_popup()`
- Kelas anak → wajib implementasi `get_info_popup()`

````python
# ============================================================
# PRAKTIKUM 03: Mendesain dan Membuat Kelas OOP
# ============================================================

from abc import ABC, abstractmethod


# ── Kelas Abstrak Lokasi (blueprint / kelas induk) ────────────────────
class Lokasi(ABC):
    """Kelas abstrak sebagai blueprint untuk semua jenis lokasi geografis."""

    def __init__(self, nama: str, latitude: float, longitude: float):
        self.nama = str(nama) if nama else 'Tanpa Nama'
        try:
            self.latitude  = float(latitude)
            self.longitude = float(longitude)
        except (ValueError, TypeError):
            print(f'  Peringatan: Koordinat tidak valid untuk "{self.nama}". Set ke (0.0, 0.0).')
            self.latitude, self.longitude = 0.0, 0.0

    def get_koordinat(self) -> tuple:
        """Mengembalikan tuple (latitude, longitude)."""
        return (self.latitude, self.longitude)

    @abstractmethod
    def get_info_popup(self) -> str:
        """Metode abstrak — WAJIB diimplementasikan di setiap kelas anak."""
        pass

    def __repr__(self) -> str:
        return (f"{type(self).__name__}(nama='{self.nama}', "
                f"lat={self.latitude:.4f}, lon={self.longitude:.4f})")

    def __str__(self) -> str:
        return f"{self.nama} [{type(self).__name__}]"


# ── Kelas Anak: TempatWisata ──────────────────────────────────────────
class TempatWisata(Lokasi):
    """Merepresentasikan tempat wisata (alam, sejarah, dll)."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 jenis: str, deskripsi: str):
        super().__init__(nama, latitude, longitude)
        self.jenis_wisata = str(jenis)     if jenis     else 'Umum'
        self.deskripsi    = str(deskripsi) if deskripsi else 'Tidak ada deskripsi.'

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>{self.jenis_wisata}</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── Kelas Anak: Kuliner ───────────────────────────────────────────────
class Kuliner(Lokasi):
    """Merepresentasikan tempat kuliner."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 menu_andalan: str):
        super().__init__(nama, latitude, longitude)
        self.menu_andalan = str(menu_andalan) if menu_andalan else 'Tidak diketahui'

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Kuliner</i><br><br>"
                f"Menu Andalan: {self.menu_andalan}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# ── Kelas Anak: TempatIbadah ──────────────────────────────────────────
class TempatIbadah(Lokasi):
    """Merepresentasikan tempat ibadah."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 agama: str = 'Umum', deskripsi: str = ''):
        super().__init__(nama, latitude, longitude)
        self.agama     = str(agama)     if agama     else 'Umum'
        self.deskripsi = str(deskripsi) if deskripsi else 'Tempat Ibadah'

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Tempat Ibadah ({self.agama})</i><br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


# --- Tes Kelas ---
print('--- Praktikum 3: Tes Kelas OOP ---\n')

tw  = TempatWisata('Lawang Sewu', -6.9840, 110.4105, 'Wisata Sejarah', 'Bangunan bersejarah.')
kl  = Kuliner('Toko Oen', -6.9715, 110.4235, 'Es Krim dan Kue Kolonial')
ti  = TempatIbadah('Masjid Agung Jawa Tengah', -6.9892, 110.4452, 'Islam', 'Masjid megah.')

for obj in [tw, kl, ti]:
    print(repr(obj))
    print(f'  Koordinat : {obj.get_koordinat()}')
    print(f'  Popup Info: {obj.get_info_popup()[:60]}...')
    print()

print('✅ Semua kelas berhasil dibuat!')
print('\n--- Praktikum 3 Selesai ---')
````

## 🔄 Praktikum 04 — Membuat Objek dari Data Pandas
**Tujuan:** Iterasi DataFrame → buat instance objek sesuai kolom `Tipe`.

Alur:
1. Baca CSV → DataFrame
2. Loop tiap baris (`iterrows()`)
3. Cek kolom `Tipe` → buat objek yang sesuai
4. Simpan ke list

````python
# ============================================================
# PRAKTIKUM 04: Membuat Objek dari Data Pandas
# ============================================================
# (Kelas Lokasi, TempatWisata, Kuliner, TempatIbadah
#  sudah terdefinisi dari sel Praktikum 03 di atas)


def buat_objek_lokasi_dari_df(dataframe) -> list:
    """
    Mengiterasi DataFrame Pandas dan membuat list berisi objek-objek
    Lokasi (atau turunannya) berdasarkan kolom 'Tipe' di setiap baris.

    Returns:
        list: List berisi instance objek Lokasi atau turunannya.
    """
    list_objek_lokasi = []

    if dataframe is None or dataframe.empty:
        print('DataFrame kosong atau None, tidak ada objek dibuat.')
        return list_objek_lokasi

    print('\nMembuat objek dari DataFrame...')

    for index, row in dataframe.iterrows():
        # Ambil data dari tiap kolom
        nama      = row.get('Nama', None)
        lat       = row.get('Latitude', None)
        lon       = row.get('Longitude', None)
        tipe      = row.get('Tipe', 'Lainnya')
        deskripsi = row.get('Deskripsi', '')

        # Validasi data wajib tidak boleh kosong
        if nama is None or lat is None or lon is None:
            print(f'  -> Melewati baris {index}: Nama/Latitude/Longitude tidak lengkap.')
            continue

        objek = None
        try:
            # Percabangan berdasarkan nilai kolom 'Tipe'
            if 'Wisata' in tipe or tipe == 'Landmark':
                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)

            elif tipe == 'Kuliner':
                # Gunakan kolom Deskripsi sebagai menu andalan
                objek = Kuliner(nama, lat, lon, deskripsi)

            elif 'Ibadah' in tipe:
                # Tentukan agama dari nama/deskripsi jika ada
                agama_info = 'Umum'
                if 'Masjid' in nama or 'Islam' in tipe:    agama_info = 'Islam'
                elif 'Gereja' in nama:                     agama_info = 'Kristen'
                elif 'Klenteng' in nama or 'Pura' in nama: agama_info = 'Tridharma'
                objek = TempatIbadah(nama, lat, lon, agama_info, deskripsi)

            else:
                print(f"  -> Peringatan: Tipe '{tipe}' untuk '{nama}' tidak dikenali.")

            if objek:
                list_objek_lokasi.append(objek)

        except Exception as e:
            print(f"  -> GAGAL membuat objek untuk '{nama}' di baris {index}: {e}")

    print(f'Total {len(list_objek_lokasi)} objek berhasil dibuat dari {len(dataframe)} baris data.')
    return list_objek_lokasi


# --- Kode Utama ---
print('--- Praktikum 4: Membuat Objek dari Data Pandas ---')

df_lokasi        = baca_data_lokasi(NAMA_FILE_CSV)       # Fungsi dari Praktikum 2
list_semua_lokasi = buat_objek_lokasi_dari_df(df_lokasi)

print('\n--- Daftar Objek Lokasi yang Berhasil Dibuat ---')
for idx, lok in enumerate(list_semua_lokasi):
    print(f'{idx+1:2}. {repr(lok)}')

print('\n--- Praktikum 4 Selesai ---')
````

## 🗺️ Praktikum 05 — Visualisasi Peta dengan Folium
**Tujuan:** Membuat peta HTML interaktif dengan marker untuk setiap objek lokasi.

- `folium.Map()` → buat peta dasar
- `folium.Marker()` → tambah penanda di peta
- `peta.save()` → simpan ke file HTML
- Popup info diambil **polimorfik** dari tiap objek via `get_info_popup()`

````python
# ============================================================
# PRAKTIKUM 05: Visualisasi Peta dengan Folium
# ============================================================

import folium

NAMA_FILE_PETA = 'peta_interaktif_semarang.html'


def buat_peta_lokasi_folium(list_objek: list, file_output: str = 'peta_lokasi.html'):
    """
    Membuat peta Folium interaktif dengan marker untuk setiap objek lokasi.

    Args:
        list_objek  : List berisi instance objek turunan Lokasi.
        file_output : Nama file HTML untuk menyimpan peta.
    """
    if not list_objek:
        print('Tidak ada objek lokasi untuk dipetakan.')
        return

    print(f'\nMemulai pembuatan peta Folium dari {len(list_objek)} lokasi...')

    # 1. Tentukan titik tengah peta (dari objek pertama)
    try:
        lat_tengah = list_objek[0].latitude
        lon_tengah = list_objek[0].longitude
    except IndexError:
        lat_tengah, lon_tengah = -6.9929, 110.4200  # Default: Semarang

    # 2. Buat objek peta Folium
    peta = folium.Map(
        location=[lat_tengah, lon_tengah],
        zoom_start=13,
        tiles='OpenStreetMap'
    )
    print(f'  -> Peta dibuat, berpusat di ({lat_tengah:.4f}, {lon_tengah:.4f})')

    # 3. Tambahkan marker untuk setiap lokasi
    jumlah_marker_valid = 0
    for lok in list_objek:
        koordinat = lok.get_koordinat()

        # Lewati jika koordinat tidak valid (0.0, 0.0 dari error)
        if koordinat == (0.0, 0.0):
            print(f"  -> Melewati marker untuk '{lok.nama}' (koordinat tidak valid).")
            continue

        # Ambil popup info secara POLIMORFIK dari setiap objek
        # TempatWisata.get_info_popup() / Kuliner.get_info_popup() / dll
        info_popup_html = lok.get_info_popup()

        folium.Marker(
            location=koordinat,
            popup=folium.Popup(info_popup_html, max_width=300),
            tooltip=lok.nama   # teks saat hover
        ).add_to(peta)

        jumlah_marker_valid += 1

    # 4. Simpan peta ke file HTML
    try:
        peta.save(file_output)
        print(f'\n✅ Peta berhasil disimpan sebagai "{file_output}".')
        print(f'   Total marker ditambahkan: {jumlah_marker_valid}')
    except Exception as e:
        print(f'\n❌ ERROR saat menyimpan peta: {type(e).__name__} - {e}')

    return peta   # kembalikan objek peta agar bisa ditampilkan di Colab


# --- Kode Utama ---
print('--- Praktikum 5: Visualisasi Peta dengan Folium ---')

peta_hasil = buat_peta_lokasi_folium(list_semua_lokasi, NAMA_FILE_PETA)

print(f'\n📌 Cara melihat peta:')
print(f'   Di Google Colab: Klik File -> klik kanan "{NAMA_FILE_PETA}" -> Download')
print(f'   Lalu buka file HTML tersebut di browser Anda.')
print('\n--- Praktikum 5 Selesai ---')
````

## 📝 Praktikum 06 — File Handling Tambahan (Menyimpan Log)
**Tujuan:** Mencatat (log) setiap proses pembuatan peta ke file teks menggunakan mode `'a'` (append).

Mode file:
- `'w'` → tulis baru (hapus isi lama)
- `'a'` → **append** (tambah di akhir, tidak hapus)
- `'r'` → baca

````python
# ============================================================
# PRAKTIKUM 06: File Handling Tambahan (Menyimpan Log)
# ============================================================

import datetime

FILE_LOG     = 'proses_peta.log'
NAMA_FILE_PETA2 = 'peta_interaktif_semarang_v2.html'


# ── Fungsi untuk Menulis Log ──────────────────────────────────────────
def tulis_log(pesan: str, file_log: str = 'proses_peta.log'):
    """
    Menulis pesan log ke file dengan timestamp.
    Menggunakan mode 'a' (append) agar log lama tidak terhapus.
    """
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    try:
        with open(file_log, 'a', encoding='utf-8') as f:
            f.write(f'[{timestamp}] {pesan}\n')
    except IOError as e:
        print(f'ERROR: Gagal menulis ke file log "{file_log}": {e}')


# ── Fungsi buat_peta_lokasi_folium (versi dengan logging) ────────────
def buat_peta_dengan_log(list_objek: list, file_output: str = 'peta_lokasi.html'):
    """
    Versi Praktikum 6: sama seperti Praktikum 5 tapi menambahkan
    pencatatan log di setiap titik penting proses.
    """
    nama_fungsi = 'buat_peta_dengan_log'

    if not list_objek:
        pesan = f'[{nama_fungsi}] GAGAL: Tidak ada data lokasi untuk dipetakan.'
        print(pesan)
        tulis_log(pesan, FILE_LOG)
        return

    # Log: mulai proses
    pesan_mulai = (f'[{nama_fungsi}] Memulai pembuatan peta "{file_output}" '
                   f'dengan {len(list_objek)} lokasi.')
    print(pesan_mulai)
    tulis_log(pesan_mulai, FILE_LOG)

    # Buat peta
    try:
        lat_tengah = list_objek[0].latitude
        lon_tengah = list_objek[0].longitude
    except IndexError:
        lat_tengah, lon_tengah = -6.9929, 110.4200

    peta = folium.Map(location=[lat_tengah, lon_tengah], zoom_start=12)

    jumlah_marker  = 0
    lokasi_dilewati = []

    for lok in list_objek:
        koordinat = lok.get_koordinat()
        if koordinat != (0.0, 0.0):
            info_popup_html = lok.get_info_popup()
            folium.Marker(
                location=koordinat,
                popup=folium.Popup(info_popup_html, max_width=300),
                tooltip=lok.nama
            ).add_to(peta)
            jumlah_marker += 1
        else:
            lokasi_dilewati.append(lok.nama)

    # Log: lokasi yang dilewati (jika ada)
    if lokasi_dilewati:
        pesan_lewat = (f'[{nama_fungsi}] Melewati marker untuk: '
                       f'{", ".join(lokasi_dilewati)} (koordinat tidak valid).')
        print(f'  -> Peringatan: {pesan_lewat}')
        tulis_log(pesan_lewat, FILE_LOG)

    # Simpan peta dan log hasilnya
    try:
        peta.save(file_output)
        pesan_sukses = (f'[{nama_fungsi}] Peta "{file_output}" berhasil dibuat '
                        f'dengan {jumlah_marker} marker.')
        print(f'\n✅ {pesan_sukses}')
        tulis_log(pesan_sukses, FILE_LOG)
    except Exception as e:
        pesan_error = (f'[{nama_fungsi}] ERROR saat menyimpan peta "{file_output}": '
                       f'{type(e).__name__} - {e}')
        print(f'\n❌ {pesan_error}')
        tulis_log(pesan_error, FILE_LOG)


# --- Kode Utama ---
print('--- Praktikum 6: File Handling Tambahan (Log) ---\n')

# Jalankan pertama kali
buat_peta_dengan_log(list_semua_lokasi, NAMA_FILE_PETA2)

# Jalankan kedua kali → log akan DITAMBAH (append), tidak ditimpa
print('\nMenjalankan lagi untuk demo mode append...')
buat_peta_dengan_log(list_semua_lokasi, 'peta_ketiga.html')

# Tampilkan isi file log
print(f'\n--- Isi File Log "{FILE_LOG}" ---')
try:
    with open(FILE_LOG, 'r', encoding='utf-8') as f:
        print(f.read())
except FileNotFoundError:
    print(f'File log belum ada.')

print('--- Praktikum 6 Selesai ---')
````

## 📌 TUGAS — Mini Project SIG (Pengembangan)

- **c)** Baca konfigurasi peta dari file `config_peta.txt`

````python
# ============================================================
# TUGAS: Mini Project SIG — Pengembangan
# ============================================================

# ── (a) KELAS BARU: Museum dan TamanKota ─────────────────────────────

class Museum(Lokasi):
    """Merepresentasikan museum."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 koleksi_utama: str, deskripsi: str = ''):
        super().__init__(nama, latitude, longitude)
        self.koleksi_utama = str(koleksi_utama) if koleksi_utama else 'Beragam'
        self.deskripsi     = str(deskripsi)     if deskripsi     else 'Museum.'

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Museum</i><br><br>"
                f"Koleksi Utama: {self.koleksi_utama}<br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


class TamanKota(Lokasi):
    """Merepresentasikan taman kota / ruang terbuka hijau."""

    def __init__(self, nama: str, latitude: float, longitude: float,
                 fasilitas: str, deskripsi: str = ''):
        super().__init__(nama, latitude, longitude)
        self.fasilitas = str(fasilitas) if fasilitas else 'Umum'
        self.deskripsi = str(deskripsi) if deskripsi else 'Taman kota.'

    def get_info_popup(self) -> str:
        return (f"<h4><b>{self.nama}</b></h4>"
                f"<i>Taman Kota</i><br><br>"
                f"Fasilitas: {self.fasilitas}<br><br>"
                f"{self.deskripsi}<br><br>"
                f"Koordinat: ({self.latitude:.4f}, {self.longitude:.4f})")


print('✅ Kelas Museum dan TamanKota berhasil didefinisikan.')
````

````python
# ── (a) Tambah 3 lokasi baru ke CSV ──────────────────────────────────

lokasi_baru = [
    'Museum Ranggawarsita,-6.9833,110.4104,Museum,Museum terbesar di Jawa Tengah dengan koleksi arkeologi dan sejarah.',
    'Taman Indonesia Kaya,-6.9905,110.4201,Taman Kota,Taman kota modern dengan panggung pertunjukan dan area bermain.',
    'Museum Rekor Dunia Indonesia,-6.9872,110.4120,Museum,Museum berisi rekor-rekor unik dari seluruh Indonesia.',
]

# Tambahkan ke file CSV menggunakan mode 'a' (append)
with open(NAMA_FILE_CSV, 'a', encoding='utf-8') as f:
    for baris in lokasi_baru:
        f.write(baris + '\n')

# Verifikasi
df_baru = pd.read_csv(NAMA_FILE_CSV)
print(f'✅ CSV diperbarui. Total baris sekarang: {len(df_baru)}')
print('\nTiga baris terakhir:')
print(df_baru.tail(3).to_string(index=False))
````

````python
# ── (a) Modifikasi fungsi buat_objek agar kenali tipe baru ───────────

def buat_objek_lokasi_dari_df_v2(dataframe) -> list:
    """
    Versi 2: sama seperti Praktikum 4 tapi menambahkan
    dukungan untuk kelas Museum dan TamanKota.
    """
    list_objek_lokasi = []

    if dataframe is None or dataframe.empty:
        print('DataFrame kosong atau None.')
        return list_objek_lokasi

    print('Membuat objek dari DataFrame (versi 2 dengan tipe baru)...')

    for index, row in dataframe.iterrows():
        nama      = row.get('Nama', None)
        lat       = row.get('Latitude', None)
        lon       = row.get('Longitude', None)
        tipe      = row.get('Tipe', 'Lainnya')
        deskripsi = row.get('Deskripsi', '')

        if nama is None or lat is None or lon is None:
            continue

        objek = None
        try:
            if 'Wisata' in tipe or tipe == 'Landmark':
                objek = TempatWisata(nama, lat, lon, tipe, deskripsi)

            elif tipe == 'Kuliner':
                objek = Kuliner(nama, lat, lon, deskripsi)

            elif 'Ibadah' in tipe:
                agama_info = 'Umum'
                if 'Masjid' in nama:    agama_info = 'Islam'
                elif 'Gereja' in nama:  agama_info = 'Kristen'
                elif 'Klenteng' in nama or 'Pura' in nama: agama_info = 'Tridharma'
                objek = TempatIbadah(nama, lat, lon, agama_info, deskripsi)

            # ── TIPE BARU ──
            elif tipe == 'Museum':
                objek = Museum(nama, lat, lon, 'Beragam Koleksi', deskripsi)

            elif 'Taman' in tipe:
                objek = TamanKota(nama, lat, lon, 'Area Hijau & Hiburan', deskripsi)

            else:
                print(f"  -> Tipe '{tipe}' untuk '{nama}' tidak dikenali.")

            if objek:
                list_objek_lokasi.append(objek)

        except Exception as e:
            print(f"  -> GAGAL membuat objek untuk '{nama}': {e}")

    print(f'Total {len(list_objek_lokasi)} objek berhasil dibuat dari {len(dataframe)} baris.')
    return list_objek_lokasi


# Buat ulang list objek dengan data + tipe baru
df_lengkap         = baca_data_lokasi(NAMA_FILE_CSV)
list_lokasi_lengkap = buat_objek_lokasi_dari_df_v2(df_lengkap)

print('\nDaftar semua objek:')
for i, lok in enumerate(list_lokasi_lengkap):
    print(f'  {i+1:2}. {repr(lok)}')
````

````python
# ── (c) Buat file konfigurasi peta ───────────────────────────────────

# Isi config_peta.txt: baris 1=latitude, baris 2=longitude, baris 3=zoom
isi_config = "-6.9929\n110.4200\n13\n"

with open('config_peta.txt', 'w', encoding='utf-8') as f:
    f.write(isi_config)

print('✅ File config_peta.txt berhasil dibuat:')
print(isi_config)
````

````python
# ── (b+c) Fungsi peta dengan marker berwarna + baca konfigurasi ───────

def baca_konfigurasi_peta(file_config: str = 'config_peta.txt'):
    """
    (c) Membaca konfigurasi peta dari file teks.
    Format: baris 1=lat, baris 2=lon, baris 3=zoom.
    Jika error, gunakan nilai default Semarang.
    """
    lat_default, lon_default, zoom_default = -6.9929, 110.4200, 13
    try:
        with open(file_config, 'r', encoding='utf-8') as f:
            baris = f.readlines()
        lat  = float(baris[0].strip())
        lon  = float(baris[1].strip())
        zoom = int(baris[2].strip())
        print(f'  -> Konfigurasi dibaca: lat={lat}, lon={lon}, zoom={zoom}')
        return lat, lon, zoom
    except FileNotFoundError:
        print(f'  -> config_peta.txt tidak ditemukan. Pakai nilai default.')
    except (ValueError, IndexError) as e:
        print(f'  -> Error membaca config: {e}. Pakai nilai default.')
    return lat_default, lon_default, zoom_default


def buat_peta_tugas(list_objek: list, file_output: str = 'peta_tugas.html',
                    file_config: str = 'config_peta.txt'):
    """
    Versi TUGAS:
    - (b) Marker berwarna berbeda per tipe objek
    - (c) Koordinat tengah & zoom dibaca dari config_peta.txt
    """
    if not list_objek:
        print('Tidak ada objek untuk dipetakan.')
        return

    print(f'\nMemulai pembuatan peta TUGAS dari {len(list_objek)} lokasi...')

    # (c) Baca konfigurasi dari file
    print('Membaca konfigurasi peta...')
    lat_tengah, lon_tengah, zoom = baca_konfigurasi_peta(file_config)

    peta = folium.Map(
        location=[lat_tengah, lon_tengah],
        zoom_start=zoom,
        tiles='OpenStreetMap'
    )

    # (b) Mapping tipe objek → warna & ikon marker
    # Warna yang didukung folium: red, blue, green, purple, orange,
    #                             darkred, lightred, beige, darkblue,
    #                             darkgreen, cadetblue, gray, lightgray, black
    def tentukan_ikon(lok):
        if isinstance(lok, TempatWisata):
            return folium.Icon(color='blue',      icon='camera',     prefix='fa')
        elif isinstance(lok, Kuliner):
            return folium.Icon(color='red',       icon='cutlery',    prefix='fa')
        elif isinstance(lok, TempatIbadah):
            return folium.Icon(color='green',     icon='star',       prefix='fa')
        elif isinstance(lok, Museum):
            return folium.Icon(color='purple',    icon='university', prefix='fa')
        elif isinstance(lok, TamanKota):
            return folium.Icon(color='darkgreen', icon='tree',       prefix='fa')
        else:
            return folium.Icon(color='gray',      icon='info-sign')

    jumlah_marker = 0
    for lok in list_objek:
        koordinat = lok.get_koordinat()
        if koordinat == (0.0, 0.0):
            continue

        folium.Marker(
            location=koordinat,
            popup=folium.Popup(lok.get_info_popup(), max_width=300),
            tooltip=f"{lok.nama} ({type(lok).__name__})",
            icon=tentukan_ikon(lok)   # (b) ikon berwarna sesuai tipe
        ).add_to(peta)
        jumlah_marker += 1

    # Tambahkan legenda warna
    legenda_html = """
    <div style="position: fixed; bottom: 30px; left: 30px; z-index:9999;
                background-color: white; padding: 10px; border-radius: 8px;
                border: 2px solid grey; font-size: 13px;">
        <b>Legenda Lokasi</b><br>
        <i class="fa fa-camera" style="color:blue"></i> Tempat Wisata<br>
        <i class="fa fa-cutlery" style="color:red"></i> Kuliner<br>
        <i class="fa fa-star" style="color:green"></i> Tempat Ibadah<br>
        <i class="fa fa-university" style="color:purple"></i> Museum<br>
        <i class="fa fa-tree" style="color:darkgreen"></i> Taman Kota
    </div>
    """
    peta.get_root().html.add_child(folium.Element(legenda_html))

    # Simpan
    try:
        peta.save(file_output)
        tulis_log(f'[buat_peta_tugas] Peta "{file_output}" dibuat dengan {jumlah_marker} marker.', FILE_LOG)
        print(f'\n✅ Peta TUGAS disimpan sebagai "{file_output}".')
        print(f'   Total marker: {jumlah_marker}')
        print('   Marker berwarna: Biru=Wisata | Merah=Kuliner | Hijau=Ibadah | Ungu=Museum | HijauTua=Taman')
    except Exception as e:
        print(f'❌ ERROR menyimpan peta: {e}')

    return peta


# --- Kode Utama Tugas ---
print('=== TUGAS: Mini Project SIG ===')
peta_tugas = buat_peta_tugas(list_lokasi_lengkap, 'peta_tugas_sig.html', 'config_peta.txt')
````
