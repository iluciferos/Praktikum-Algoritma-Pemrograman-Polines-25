### Praktikum 04 – Fungsi dengan *args dan **kwargs

```python
def jumlahkan(*args):
    return sum(args)
def biodata (**kwargs):
    for key, value in kwargs.items():
        print(f"{key}  : {value}")

print("Total : ", jumlahkan(1,2,3,4,5))
biodata(Nama="Budi", Umur="20", Prodi="TRK")
```

### Praktikum 07 – Fungsi Decorator

```python
def decorator_sapa(fungsi):
    def wrapper():
        print("Sebelum fungsi dijalankan...")
        fungsi()
        print("Setelah fungsi dijalankan...")
    return wrapper

@decorator_sapa
def sapa():
    print("Halo, Mahasiswa Polines!")
sapa()
```

```python
# --- d. Lambda untuk menghitung nilai akhir ---
# Lambda menerima dua argumen (nilai_tugas, nilai_ujian) dan menghitung nilai akhir berdasarkan bobot.
# Bobot: 40% Tugas, 60% Ujian
hitung_nilai_akhir = lambda nilai_tugas, nilai_ujian: (0.4 * nilai_tugas) + (0.6 * nilai_ujian)


# --- a. Fungsi untuk menerima input nilai mahasiswa ---
def input_nilai():
    """
    Menerima input nilai dari pengguna dan menyimpannya dalam sebuah list.
    Pengguna dapat berhenti memasukkan nilai dengan mengetik 'selesai'.
    """
    daftar_nilai = []
    print("--- Input Nilai Mahasiswa ---")
    print("Masukkan nilai satu per satu. Ketik 'selesai' jika sudah.")
    while True:
        masukan = input("Masukkan nilai: ")
        if masukan.lower() == 'selesai':
            break
        try:
            # Konversi input ke float dan tambahkan ke list
            nilai = float(masukan)
            daftar_nilai.append(nilai)
        except ValueError:
            # Handle jika input bukan angka
            print("Input tidak valid. Silakan masukkan angka.")
    return daftar_nilai


# --- b. Fungsi untuk menghitung rata-rata ---
def hitung_rata(daftar_nilai):
    """
    Menghitung rata-rata dari sebuah list nilai.
    Mengembalikan 0 jika list kosong untuk menghindari pembagian dengan nol.
    """
    if not daftar_nilai:  # Cek apakah list kosong
        return 0
    total = sum(daftar_nilai)
    rata_rata = total / len(daftar_nilai)
    return rata_rata


# --- c. Fungsi untuk menentukan status kelulusan ---
def status_kelulusan(rata):
    """
    Menentukan status kelulusan berdasarkan nilai rata-rata.
    Kriteria lulus: rata-rata >= 60.
    """
    if rata >= 60:
        return "Lulus"
    else:
        return "Tidak Lulus"


# --- Alur Program Utama ---
if __name__ == "__main__":
    # 1. Memanggil fungsi input_nilai untuk mendapatkan daftar nilai
    nilai_mahasiswa = input_nilai()

    # 2. Memeriksa apakah ada nilai yang dimasukkan
    if not nilai_mahasiswa:
        print("\nTidak ada nilai yang dimasukkan. Program berhenti.")
    else:
        # 3. Menghitung rata-rata dari nilai yang telah diinput
        nilai_rata_rata = hitung_rata(nilai_mahasiswa)
        print(f"\nDaftar Nilai: {nilai_mahasiswa}")
        print(f"Rata-rata Nilai: {nilai_rata_rata:.2f}")

        # 4. Menentukan status kelulusan berdasarkan rata-rata
        status = status_kelulusan(nilai_rata_rata)
        print(f"Status Kelulusan: {status}")

        # --- Bagian Tambahan: Menggunakan Lambda ---
        print("\n--- Menghitung Nilai Akhir dengan Bobot ---")
        try:
            # Meminta input nilai tugas dan ujian
            input_tugas = float(input("Masukkan nilai Tugas (0-100): "))
            input_ujian = float(input("Masukkan nilai Ujian (0-100): "))

            # Menghitung nilai akhir menggunakan fungsi lambda
            nilai_akhir = hitung_nilai_akhir(input_tugas, input_ujian)
            print(f"Nilai Akhir (40% Tugas + 60% Ujian): {nilai_akhir:.2f}")

        except ValueError:
            print("Input untuk nilai tugas atau ujian tidak valid.")
```
